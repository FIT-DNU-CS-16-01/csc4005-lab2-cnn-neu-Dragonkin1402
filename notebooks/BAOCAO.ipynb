## 4. Phân tích và Kết luận

### Khi nào nên dùng Transfer Learning?
* **Dữ liệu nhỏ:** NEU-CLS có số lượng ảnh hạn chế, Transfer Learning giúp mô hình hội tụ nhanh mà không cần hàng ngàn ảnh mỗi lớp[cite: 1].
* **Độ chính xác cao:** Các đặc trưng hình học của lỗi thép (vết nứt, chấm đen) tương đồng với các đặc trưng cạnh/góc mà ResNet đã học được trước đó[cite: 1].

### Khi nào dùng mô hình From Scratch?
* Khi bài toán có dữ liệu cực kỳ đặc thù (không phải ảnh quang học thông thường) hoặc khi cần một mô hình siêu nhẹ để chạy trên các thiết bị nhúng có phần cứng yếu[cite: 1].
